# 🔍 Détection des doublons utilisateurs – Projet ETL

Ce notebook permet de :
- Importer les données Octopus
- Nettoyer les noms utilisateurs
- Détecter les doublons **exacts**
- Détecter les doublons **similaires** (fuzzy matching)
- Calculer des statistiques et afficher des graphiques
- Exporter les résultats


## 1. Importer les librairies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import unicodedata
import os, sys

# Installation automatique de rapidfuzz si absent
try:
    from rapidfuzz import fuzz, process
except ImportError:
    import subprocess
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'rapidfuzz', '-q'])
    from rapidfuzz import fuzz, process

%matplotlib inline
print('✅ Librairies chargées avec succès')
```

## 2. Configuration – Paramètres à modifier

```python
# ============================================================
# 🔧 PARAMÈTRES – À adapter selon votre environnement
# ============================================================

INPUT_FILE           = r"C:\Users\Bouchra\OneDrive\Documents\Data accounts.xlsx"
NAME_COLUMN          = 'Full Name'
SIMILARITY_THRESHOLD = 90   # Entre 0 et 100
OUTPUT_FOLDER        = r"C:\Users\Bouchra\OneDrive\Documents"

# ============================================================
print(f'📂 Fichier source   : {INPUT_FILE}')
print(f'🎯 Seuil similarité : {SIMILARITY_THRESHOLD}%')
print(f'📁 Dossier sortie   : {OUTPUT_FOLDER}')
```

## 3. Charger le fichier Excel

```python
try:
    df_raw = pd.read_excel(INPUT_FILE)
    print(f'✅ Fichier chargé : {df_raw.shape[0]} lignes, {df_raw.shape[1]} colonnes')
    print(f'📋 Colonnes disponibles : {list(df_raw.columns)}')
except FileNotFoundError:
    raise FileNotFoundError(
        f"❌ Fichier introuvable : {INPUT_FILE}\n"
        "Vérifiez le chemin dans la section Configuration."
    )

if NAME_COLUMN not in df_raw.columns:
    raise ValueError(
        f"❌ Colonne '{NAME_COLUMN}' absente.\n"
        f"Colonnes disponibles : {list(df_raw.columns)}"
    )

df = df_raw[[NAME_COLUMN]].copy()
df.head()
```

## 4. Nettoyage des noms

```python
def clean_name(name):
    """Normalise un nom : minuscules, sans accents, sans espaces superflus."""
    if pd.isna(name) or str(name).strip() == '':
        return None  # None pour pouvoir filtrer avec dropna()
    name = str(name).lower().strip()
    name = unicodedata.normalize('NFKD', name).encode('ASCII', 'ignore').decode('utf-8')
    name = ' '.join(name.split())  # normalise les espaces multiples
    return name

df['clean_name'] = df[NAME_COLUMN].apply(clean_name)

n_null = df['clean_name'].isna().sum()
print(f'⚠️  Noms vides / NaN supprimés : {n_null}')
print(f'✅ Noms valides conservés      : {df["clean_name"].notna().sum()}')
df.head(10)
```

## 5. Détection des doublons exacts

```python
exact_mask = df.duplicated('clean_name', keep=False) & df['clean_name'].notna()
exact_duplicates_rows = df[exact_mask].copy()

# Tableau récapitulatif : nom → nombre d'occurrences + exemples de graphie
exact_summary = (
    df[df['clean_name'].notna()]
    .groupby('clean_name')[NAME_COLUMN]
    .agg(Count='count', Exemples=lambda x: ' | '.join(x.unique()[:3]))
    .reset_index()
    .query('Count > 1')
    .sort_values('Count', ascending=False)
    .rename(columns={'clean_name': 'Nom nettoyé'})
)

print(f'🔴 Noms distincts en doublon exact : {len(exact_summary)}')
print(f'🔴 Lignes concernées               : {exact_duplicates_rows.shape[0]}')
exact_summary.head(20)
```

## 6. Détection des doublons similaires (fuzzy matching)

```python
names = df['clean_name'].dropna().unique().tolist()
print(f'🔎 Comparaison fuzzy sur {len(names)} noms uniques (seuil = {SIMILARITY_THRESHOLD}%)...')

similar_duplicates = []

# process.extract est vectorisé → bien plus rapide que la double boucle naïve
for i, name in enumerate(names):
    matches = process.extract(
        name,
        names[i + 1:],          # uniquement les noms suivants (évite les doublons de paires)
        scorer=fuzz.ratio,
        score_cutoff=SIMILARITY_THRESHOLD
    )
    for match_name, score, _ in matches:
        similar_duplicates.append({
            'Nom 1': name,
            'Nom 2': match_name,
            'Score similarité (%)': score
        })

similar_df = pd.DataFrame(similar_duplicates).sort_values('Score similarité (%)', ascending=False)
print(f'🟠 Paires similaires trouvées : {len(similar_df)}')
similar_df.head(20)
```

## 7. Statistiques

```python
total_rows      = len(df)
total_unique    = df['clean_name'].nunique()
n_exact_names   = len(exact_summary)
n_exact_rows    = exact_duplicates_rows.shape[0]
n_similar_pairs = len(similar_df)
pct_exact       = (n_exact_rows / total_rows * 100) if total_rows else 0

print('=' * 45)
print('         RAPPORT – DOUBLONS OCTOPUS')
print('=' * 45)
print(f'  Total lignes chargées   : {total_rows:>6}')
print(f'  Noms uniques (nettoyés) : {total_unique:>6}')
print(f'  Noms en doublon exact   : {n_exact_names:>6}')
print(f'  Lignes dupliquées       : {n_exact_rows:>6}  ({pct_exact:.1f}%)')
print(f'  Paires similaires       : {n_similar_pairs:>6}  (seuil {SIMILARITY_THRESHOLD}%)')
print('=' * 45)
```

## 8. Visualisation des doublons

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# --- Camembert : doublons exacts ---
sizes_pie  = [n_exact_rows, total_rows - n_exact_rows]
labels_pie = [f'Doublons\n({n_exact_rows})', f'Uniques\n({total_rows - n_exact_rows})']
axes[0].pie(sizes_pie, labels=labels_pie, colors=['#e74c3c', '#2ecc71'],
            autopct='%1.1f%%', startangle=90)
axes[0].set_title('Doublons exacts vs Uniques', fontsize=13, fontweight='bold')

# --- Barres horizontales : Top 10 noms les plus dupliqués ---
top10 = exact_summary.head(10)
if not top10.empty:
    axes[1].barh(top10['Nom nettoyé'], top10['Count'], color='#e67e22')
    axes[1].invert_yaxis()
    axes[1].set_xlabel("Nombre d'occurrences")
    axes[1].set_title('Top 10 des noms les plus dupliqués', fontsize=13, fontweight='bold')
else:
    axes[1].text(0.5, 0.5, 'Aucun doublon détecté', ha='center', va='center')
    axes[1].set_title('Top 10 doublons', fontsize=13)

plt.tight_layout()
plt.show()
```

## 9. Exporter les résultats

```python
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

exact_path   = os.path.join(OUTPUT_FOLDER, 'exact_duplicates.csv')
similar_path = os.path.join(OUTPUT_FOLDER, 'similar_duplicates.csv')

# encoding='utf-8-sig' → lisible directement dans Excel (Windows)
exact_summary.to_csv(exact_path,   index=False, encoding='utf-8-sig')
similar_df.to_csv(similar_path, index=False, encoding='utf-8-sig')

print('✅ Exports réussis :')
print(f'   📄 {exact_path}   ({len(exact_summary)} lignes)')
print(f'   📄 {similar_path} ({len(similar_df)} lignes)')
```
